```r
# Counts used in Sections 3.7-3.9
list.files()

install.packages("openxlsx")   # only needed the first time
library(openxlsx)
```

```r
# 1. Read the merged CARD annotation table (sheet "Merged")
card <- read.xlsx("Final_CARD_merged (3)2.xlsx", sheet = "Merged")
# Keep only the gene rows (the colour-key rows at the bottom have no match type)
card <- card[!is.na(card$Match.type), ]
stopifnot(length(unique(card$AMRFinderPlus.Gene)) == 128)
```

```r
# 2. The 29 genes that were not matched at gene level (Appendix A)
not_matched <- c("aac(6')", "bla", "blaMCA", "blaOKP-B", "blaOXA-10",
  "blaPSE", "blaRSD2", "blaRTG", "blaSHV", "blaTEM-1", "blaVEB",
  "blaVEB-1", "catA16", "dfrA48", "lnu(AN2)", "lnu(I)", "mcr-5",
  "mef(A)", "mrx(A)", "msr", "qnrS", "qnrVC", "tet", "tet(X2)",
  "tmexC", "tmexD", "tmexD3", "toprJ", "toprJ1")
c99 <- card[!(card$AMRFinderPlus.Gene %in% not_matched), ]
stopifnot(length(unique(c99$AMRFinderPlus.Gene)) == 99)
```

```r
# 3. Section 3.7: number of genes per drug class and per mechanism
cls <- unlist(strsplit(c99$Drug.Class, ";\\s*"))
print(sort(table(cls), decreasing = TRUE))
mech <- unlist(strsplit(c99$Resistance.Mechanism, ";\\s*"))
print(sort(table(mech), decreasing = TRUE))
print(round(99 / 128 * 100, 1))    # percentage of matched genes
```

```r
# 4. Sections 3.8-3.9: presence/absence of the matched genes
m <- read.csv("arg_presence_absence.csv", check.names = FALSE)
m99 <- m[m$Gene %in% c99$AMRFinderPlus.Gene, ]
print(table(rowSums(m99[, -1])))   # number of genes found in 1-5 samples
print(colSums(m99[, -1]))          # number of matched genes per sample
```

```r
# 5. Genes found in four samples, and genes found in only one sample (by sample)
n99 <- rowSums(m99[, -1])
print(m99$Gene[n99 == 4])
print(colSums(m99[n99 == 1, -1]))
```

```r
# 6. All 128 genes
n128 <- rowSums(m[, -1])
print(table(n128))
print(colSums(m[n128 == 1, -1]))
```

```r
# 7. Number of samples in which each unmatched gene was detected
nm <- m[m$Gene %in% not_matched, ]
print(setNames(rowSums(nm[, -1]), nm$Gene))
```
